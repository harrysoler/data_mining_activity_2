## Parte 1 · Entorno y datos de clientes
Simulamos clientes de un centro comercial con dos variables: ingreso anual y puntaje de gasto.

```python
# Qué hace esta celda: fabricar 200 clientes de mentiras (pero realistas) para practicar

import pandas as pd, numpy as np                 # pandas = tablas; numpy = cálculos
import matplotlib.pyplot as plt, seaborn as sns  # los graficadores
sns.set_style('whitegrid')                       # estilo: fondo blanco con cuadrícula
np.random.seed(42)                               # semilla fija: a todos nos salen los MISMOS clientes

# La idea: 5 "barrios" de clientes. Cada pareja (cx, cy) es el centro de un barrio:
# (ingreso típico, gasto típico). Ejemplo: (25, 20) = gana poco y gasta poco.
centros = [(25,20),(30,80),(55,50),(80,20),(85,80)]

filas = []                                       # lista vacía: aquí iremos metiendo clientes
for cx, cy in centros:                           # para cada barrio...
    for _ in range(40):                          # ...crea 40 clientes (el _ = "no me importa el contador")
        # randn() da un numerito al azar alrededor de 0; multiplicado por 6 u 8
        # dispersa a cada cliente alrededor del centro de su barrio (nadie es idéntico)
        filas.append([cx + np.random.randn()*6, cy + np.random.randn()*8])

df = pd.DataFrame(filas, columns=['ingreso','gasto'])   # convertir la lista en tabla
print('Clientes:', df.shape)                            # (filas, columnas) = (200, 2)
df.head()                                               # las primeras 5 filas, para ver cómo quedó
```

### 1.1 Verlos sin agrupar (¿ves grupos a simple vista?)

```python
# Qué hace esta celda: dibujar los 200 clientes SIN decirle nada a nadie

plt.figure(figsize=(7,5))                        # lienzo de 7x5 pulgadas
# scatter dibuja un punto por cliente: x = su ingreso, y = su gasto.
# c='gray' = todos del mismo color (todavía NO hay grupos); alpha=0.6 = puntos semitransparentes
plt.scatter(df['ingreso'], df['gasto'], c='gray', alpha=0.6)
plt.xlabel('Ingreso anual (miles)'); plt.ylabel('Puntaje de gasto')
plt.title('Clientes SIN agrupar')
plt.show()
```

**Cómo leer este gráfico:** cada punto gris es un cliente: a la derecha ganan más, arriba
gastan más. Entrecierra los ojos: ¿ves como 5 nubes de puntos? Tu cerebro acaba de hacer
clustering a ojo. El punto de la clase es que K-Means haga lo mismo, pero con matemática
y con datos que no se pueden dibujar.

### Para entenderlo fácil: K-Means es el organizador de la fiesta

Llegas a un salón con 200 invitados regados y te piden armar 5 mesas. Haces esto:

1. Pones las 5 mesas en cualquier parte.
2. Cada invitado se sienta en la mesa MÁS CERCANA.
3. Miras cada mesa y la mueves al CENTRO de los que se sentaron en ella.
4. Como las mesas se movieron, algunos invitados ahora quedan más cerca de otra mesa:
   se cambian. Vuelves al paso 3.
5. Repites hasta que las mesas dejan de moverse. Listo: 5 grupos naturales.

Eso, literalmente, es K-Means. Las "mesas" se llaman **centroides**, y el algoritmo hace
los pasos 2-3-4 solo, en milésimas de segundo. Tú solo decides CUÁNTAS mesas (el K).

---
## Parte 2 · Agrupar con K-Means
**Regla de oro:** normalizar antes, porque K-Means mide distancias.

```python
# Qué hace esta celda: pedirle a K-Means que encuentre los grupos SOLO

from sklearn.cluster import KMeans               # el agrupador
from sklearn.preprocessing import StandardScaler # el normalizador de siempre

X = df[['ingreso','gasto']]                      # doble corchete = mini tabla con esas 2 columnas

# PASO OBLIGATORIO: normalizar. ¿Por qué? K-Means agrupa por DISTANCIA, y si una
# columna tiene números grandes (ingreso en miles) y otra pequeños, la grande manda
# y la pequeña se vuelve invisible. StandardScaler las pone en igualdad (media 0, desv 1).
Xs = StandardScaler().fit_transform(X)

# KMeans con sus perillas:
#   n_clusters=5  -> queremos 5 grupos (¿por qué 5? más abajo aprendemos a elegirlo)
#   random_state=42 -> las mesas iniciales se ponen igual para todos (reproducible)
#   n_init=10     -> repite el juego 10 veces desde arranques distintos y se queda con el mejor
km = KMeans(n_clusters=5, random_state=42, n_init=10)

# fit_predict hace DOS cosas de una: aprende los grupos (fit) y le asigna a cada
# cliente su número de grupo (predict). Guardamos esa etiqueta como columna nueva.
df['grupo'] = km.fit_predict(Xs)
df['grupo'].value_counts()                       # cuántos clientes quedaron en cada grupo
```

### 2.1 El momento revelador: ver los grupos

```python
# Qué hace esta celda: el momento revelador — ver los grupos con colores (y sus centros)

plt.figure(figsize=(7,5))
# La diferencia con el dibujo anterior es UNA cosa: c=df['grupo'].
# Ahora el color de cada punto es su número de grupo; cmap='viridis' es la paleta.
plt.scatter(df['ingreso'], df['gasto'], c=df['grupo'], cmap='viridis', alpha=0.8)

# EXTRA: dibujar las MESAS (los centroides). km.cluster_centers_ las tiene, pero
# en la escala normalizada; inverse_transform las devuelve a pesos y gastos reales.
centros_reales = StandardScaler().fit(X).inverse_transform(km.cluster_centers_)
plt.scatter(centros_reales[:,0], centros_reales[:,1],
            marker='X', s=250, c='red', edgecolors='black', label='centroides (las mesas)')

plt.xlabel('Ingreso anual (miles)'); plt.ylabel('Puntaje de gasto')
plt.title('Clientes agrupados por K-Means')
plt.colorbar(label='grupo')                      # la barrita que traduce color -> número de grupo
plt.legend()
plt.show()
```

**Cómo leer este gráfico:** son los MISMOS puntos de antes, pero ahora cada color es un
grupo que K-Means encontró SOLO — nadie le dijo dónde estaban las nubes. La barra lateral
traduce color a número de grupo. Ojo con lo que viene: el número de grupo (0, 1, 2...) es
una etiqueta sin significado; el significado se lo pones tú en la siguiente sección.

### Interpreta los grupos (esto es lo valioso)
Ponle nombre de negocio a cada grupo. Por ejemplo:
- Ingreso alto + gasto alto → **clientes VIP**
- Ingreso alto + gasto bajo → **cautelosos** (oportunidad de fidelizar)
- Ingreso bajo + gasto alto → **entusiastas**

El modelo encontró los grupos; TÚ les das significado.

## Parte 3 · ¿Cuántos grupos? El método del codo

```python
# Qué hace esta celda: probar K de 1 a 10 y medir qué tan "apretado" queda cada agrupamiento

errores = []                                     # lista vacía para ir guardando la nota de cada K
for k in range(1, 11):                           # range(1, 11) = los números del 1 al 10
    km_k = KMeans(n_clusters=k, random_state=42, n_init=10).fit(Xs)   # agrupar con ESE k
    # inertia_ = la suma de distancias de cada cliente a su mesa. Chiquito = mesas bien puestas.
    errores.append(km_k.inertia_)

plt.figure(figsize=(7,4))
# 'o-' significa: pon un puntito en cada valor Y únelos con línea
plt.plot(range(1,11), errores, 'o-', color='#7C3AED')

# EXTRA: ponerle su valor a cada puntico (queda muy bien en el informe).
# annotate le pone una "notica" a un punto: QUÉ escribir, DÓNDE anclarla,
# y cuánto correrla para no tapar el punto (aquí, 8 puntos hacia arriba).
# zip empareja cada K con su error, y el for les pone la notica una por una.
for k, e in zip(range(1,11), errores):
    plt.annotate(f'{e:.0f}', (k, e), textcoords='offset points',
                 xytext=(0, 8), ha='center', fontsize=9)

plt.xlabel('Número de grupos (K)'); plt.ylabel('Error interno (inertia)')
plt.title('Método del codo — busca dónde se "dobla" la curva')
plt.show()

# PRUEBA A CAMBIAR: para el informe queda elegante señalar SOLO el codo con una flecha:
# plt.annotate('el codo', (5, errores[4]), xytext=(20, 20),
#              textcoords='offset points', arrowprops=dict(arrowstyle='->'))
```

### Cómo leer el codo

**El eje X** es cuántas mesas probamos (K) y **el eje Y** qué tan lejos quedaron los
invitados de su mesa (el error interno o inertia). Con 1 mesa el error es enorme; al
agregar mesas baja rapidísimo... hasta K=5, donde la curva se DOBLA como un codo y se
aplana: de ahí en adelante, agregar mesas casi no mejora nada. Ese doblez señala el K
ideal — y coincide con las 5 nubes que viste a ojo al principio, buena señal. En tu
actividad, esta gráfica va en el informe junto con el Silhouette.

## Parte 4 · Reglas de asociación: ¿qué se compra junto?
Analizamos tickets de compra para encontrar productos que van juntos.

### Para entenderlo fácil: Apriori es el tendero observador

El tendero de tu barrio no sabe estadística, pero después de años atendiendo sabe que
"el que lleva pan casi siempre lleva mantequilla" y por eso las pone cerca. Apriori es
ese tendero convertido en algoritmo: revisa miles de carritos y descubre qué productos
van juntos, con tres números por regla:

- **Soporte:** qué tan común es el combo ("el 40% de los carritos lleva pan y mantequilla").
- **Confianza:** qué tan fiable es la regla ("DE LOS QUE llevan pan, el 70% lleva mantequilla").
- **Lift:** si la relación es real o casualidad ("llevar pan hace 1.8 veces MÁS probable
  llevar mantequilla que al azar"). Lift = 1 es casualidad; mayor que 1 es lo interesante.

El caso legendario: una cadena descubrió que pañales y cerveza se compraban juntos
(papás jóvenes de compras). Nadie lo habría adivinado; los carritos lo contaron.

```python
# mlxtend es la librería de las reglas de asociación (no viene con las de siempre).
# Solo la primera vez en cada computador; el % es obligatorio dentro del notebook.
%pip install mlxtend --quiet
```

```python
# Qué hace esta celda: fabricar 200 carritos de mercado con costumbres escondidas

from mlxtend.frequent_patterns import apriori, association_rules

productos = ['pan','leche','mantequilla','mermelada','cafe','azucar']
np.random.seed(1)                                # semilla: mismos carritos para todos

carritos = []
for _ in range(200):                             # 200 carritos
    # {p: 0 for p in productos} crea un carrito vacío: cada producto en 0 (no lo llevó)
    fila = {p: 0 for p in productos}
    # np.random.rand() da un número al azar entre 0 y 1; "< 0.6" = pasa el 60% de las veces.
    # Así sembramos costumbres: el 60% lleva pan; DE LOS QUE llevan pan, el 70% lleva
    # mantequilla; de esos, el 60% mermelada. Y aparte: café, y con café azúcar.
    if np.random.rand()<0.6: fila['pan']=1
    if fila['pan'] and np.random.rand()<0.7: fila['mantequilla']=1
    if fila['mantequilla'] and np.random.rand()<0.6: fila['mermelada']=1
    if np.random.rand()<0.5: fila['cafe']=1
    if fila['cafe'] and np.random.rand()<0.7: fila['azucar']=1
    if np.random.rand()<0.4: fila['leche']=1
    carritos.append(fila)

# astype(bool) convierte los 0/1 en Falso/Verdadero (el formato que apriori exige)
cesta = pd.DataFrame(carritos).astype(bool)
cesta.head()                                     # cada fila = un carrito; cada columna = ¿llevó ese producto?

# La gracia: Apriori NO sabe que sembramos esas costumbres. Veamos si las descubre solo.
```

```python
# Qué hace esta celda: soltar al detective Apriori sobre los carritos

# apriori encuentra los CONJUNTOS FRECUENTES: combos de productos que aparecen
# juntos en al menos el 5% de los carritos (min_support=0.05).
# use_colnames=True: que diga "pan" y no "columna 0".
frecuentes = apriori(cesta, min_support=0.05, use_colnames=True)

# association_rules convierte esos combos en REGLAS del tipo "si lleva X -> lleva Y",
# quedándose con las que tienen lift de al menos 1 (interesantes o mejor)
reglas = association_rules(frecuentes, metric='lift', min_threshold=1)

# Nos quedamos con las 5 columnas que importan y ordenamos por lift (las más fuertes arriba)
reglas = reglas[['antecedents','consequents','support','confidence','lift']]
reglas.sort_values('lift', ascending=False).head(8)   # el top 8
```

### Cómo leer una regla
`{pan} → {mantequilla}` con confidence 0.7 y lift 1.8 significa:
- **Confidence 0.7:** el 70% de quienes compran pan también compran mantequilla.
- **Lift 1.8:** comprar pan hace **1.8 veces más probable** comprar mantequilla que al azar. ¡Relación real!

**Uso en negocio:** poner pan y mantequilla cerca, o hacer combos.

## PREPARACIÓN PARA LA ACTIVIDAD (Sección 2) · Silhouette + RFM

La actividad **"Segmentación estratégica de mercado y análisis de cestas"** (50% · cierra el **domingo 6 de septiembre, 11:59 pm**) pide en su **Fase I**: construir **RFM** (o matriz usuario-producto), visualizar con **PCA**, elegir K con el **codo Y el Silhouette Score**, y aplicar **Apriori** por segmento sobre **Online Retail**. Aquí están las tres piezas clave.

### 1) Silhouette Score — confirma cuántos grupos

```python
# Qué hace esta celda: pedir una segunda opinión sobre cuántos grupos (Silhouette)

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

Xs = StandardScaler().fit_transform(df[['ingreso','gasto']])   # normalizar (siempre)

# El Silhouette mide, para cada cliente, qué tan CÓMODO está en su mesa:
# cerca de los de su grupo y lejos de la mesa vecina. Va de -1 (mesa equivocada)
# a +1 (perfectamente ubicado). El promedio de todos es el Silhouette Score.
for k in range(2, 7):                            # probar K de 2 a 6 (con K=1 no tiene sentido)
    labels = KMeans(n_clusters=k, n_init=10, random_state=42).fit_predict(Xs)
    print(f'K={k}  Silhouette = {silhouette_score(Xs, labels):.3f}')

# El K con el Silhouette más ALTO es el mejor candidato. Tu actividad pide
# elegir K con el codo Y el Silhouette "coordinadamente": los dos deben apuntar
# al mismo número (aquí, al 5). Si no coinciden, se explica cuál pesó más y por qué.
```

### 2) RFM — resumir a cada cliente en 3 números
RFM convierte el historial de compras en 3 indicadores por cliente. Sobre Online Retail sería así:

```python
# Qué hace esta celda: convertir un historial de compras en la tabla RFM

import pandas as pd
# Datos de juguete: 6 compras de 3 clientes (cliente, cuándo compró, cuánto gastó)
tx = pd.DataFrame({
    'cliente':[1,1,2,3,3,3],
    'fecha': pd.to_datetime(['2024-01-01','2024-03-01','2024-02-01','2024-01-15','2024-04-01','2024-05-01']),
    'monto':[100,200,50,300,150,120]})           # pd.to_datetime convierte texto en FECHAS de verdad

hoy = tx['fecha'].max()                          # .max() = la fecha más reciente (nuestro "hoy")

# groupby('cliente') = "haz montoncitos, uno por cliente". agg = "de cada montoncito, calcula:"
rfm = tx.groupby('cliente').agg(
    # Recencia: días desde su ÚLTIMA compra. El lambda es una mini fórmula:
    # "toma las fechas de este cliente (f), busca la última (f.max()), réstasela a hoy, y dame los días"
    Recencia=('fecha', lambda f: (hoy - f.max()).days),
    Frecuencia=('fecha', 'count'),               # cuántas compras hizo ('count' = contar)
    Monetario=('monto', 'sum'))                  # cuánto gastó en total ('sum' = sumar)
print(rfm)

# Resultado: cada cliente resumido en 3 números. Léelo: el cliente 3 compró hace
# poco, 3 veces, y gastó 570 — cliente estrella. El 2 lleva tiempo perdido y gastó 50.
# Con esta tabla, el K-Means agrupa CLIENTES igual que agrupó los del centro comercial.
```

### 3) PCA — ver los clusters en 2D
La actividad pide **reducción de dimensionalidad con PCA** para visualizar los datos en dos dimensiones. La idea: RFM vive en 3 dimensiones (y una matriz usuario-producto en cientos); PCA las "aplasta" a 2 ejes que conservan la mayor varianza posible, para poder **graficar** los segmentos.

### Para entenderlo fácil: PCA es elegir la mejor sombra

Pon tu mano frente a una lámpara: la sombra en la pared es una versión en 2D de un objeto
en 3D. Se pierde algo (la profundidad), pero si giras la mano hasta el mejor ángulo, la
sombra conserva la forma y cualquiera reconoce que es una mano.

PCA hace exactamente eso con datos: tu tabla RFM vive en 3 dimensiones (y una matriz
usuario-producto en cientos), imposibles de dibujar. PCA "gira la lámpara" hasta el ángulo
donde la sombra en 2D conserva la MAYOR información posible, y te dice cuánta conservó
(la varianza explicada). Por eso puedes graficar tus segmentos en un plano sin mentir.

```python
# Qué hace esta celda: aplastar las 3 dimensiones del RFM a 2, para poder DIBUJAR

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

rfm_s = StandardScaler().fit_transform(rfm)      # SIEMPRE escalar antes de PCA (mismas razones que K-Means)

pca = PCA(n_components=2)                        # "quiero quedarme con 2 dimensiones"
rfm_2d = pca.fit_transform(rfm_s)                # aprender el mejor ángulo y proyectar

# explained_variance_ratio_ dice qué porcentaje de la información original
# conserva cada uno de los 2 ejes nuevos. La suma es el total retenido.
print('Varianza explicada por componente:', pca.explained_variance_ratio_.round(3))
print('Varianza total retenida:', pca.explained_variance_ratio_.sum().round(3))

# rfm_2d[:,0] = todas las filas, eje nuevo 1; rfm_2d[:,1] = eje nuevo 2
plt.scatter(rfm_2d[:,0], rfm_2d[:,1], s=80)

# EXTRA: etiquetar cada punto con SU cliente (el mismo truco annotate del codo).
# rfm.index trae los números de cliente; enumerate nos da la posición de cada uno.
for i, cliente in enumerate(rfm.index):
    plt.annotate(f'cliente {cliente}', (rfm_2d[i,0], rfm_2d[i,1]),
                 textcoords='offset points', xytext=(0, 10), ha='center', fontsize=9)

plt.xlabel('Componente principal 1'); plt.ylabel('Componente principal 2')
plt.title('Clientes proyectados a 2D con PCA')
plt.show()

# En tu actividad: colorea estos puntos con las etiquetas del K-Means para VER los segmentos.
# (Con miles de clientes NO se etiqueta punto por punto — ahí el color por segmento es la etiqueta.)
```

**Cómo leer este gráfico:** cada punto es un cliente, pero los ejes ya no son "ingreso" o
"gasto": son los dos ángulos de sombra que PCA eligió (componente 1 y 2). Las distancias
SÍ significan: puntos cercanos son clientes parecidos en su RFM completo. Con solo 3
clientes de juguete se ve poco; con los miles de Online Retail, aquí es donde tus
segmentos se ven como islas de colores. Y el número clave para tu informe salió impreso
arriba: la varianza retenida.
